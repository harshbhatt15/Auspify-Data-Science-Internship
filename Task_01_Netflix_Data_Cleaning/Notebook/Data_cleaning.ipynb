```python
df_clean = df.copy()
```

```python
# 1. Remove extra whitespace
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].str.strip()

# 2. Convert placeholder values to NaN
df_clean.replace('Not Given', np.nan, inplace=True)

# 3. Convert date_added into datetime
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'],
    format='%m/%d/%Y',
    errors='coerce'
)

# 4. Extract duration information
df_clean[['duration_value', 'duration_unit']] = (
    df_clean['duration'].str.extract(r'(\d+)\s*(.*)')
)

df_clean['duration_value'] = pd.to_numeric(
    df_clean['duration_value']
)

# 5. Remove duplicate content records
df_clean = df_clean.drop_duplicates(
    subset=df_clean.columns.drop('show_id').tolist()
)

df_clean.reset_index(drop=True, inplace=True)
```

```python
print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Rows:",
      df_clean.duplicated().sum())

print("\nData Types:")
print(df_clean.dtypes)
```

During data quality analysis, four duplicate content records were identified. Initially, three duplicate pairs were detected. After standardizing whitespace in the title column, one additional duplicate pair was discovered. Duplicate records were removed while retaining the first occurrence and preserving the original show ID.
